```python
"""
╔══════════════════════════════════════════════════════════════════════════════╗
║  RBI MONETARY POLICY SIGNALING — COMPLETE ANNOTATED PIPELINE               ║
║                                                                             ║
║  LOCAL SETUP (run once in terminal before running this file):              ║
║    pip install nltk scikit-learn statsmodels pandas scipy matplotlib       ║
║               pdfplumber yfinance pandas-datareader openpyxl               ║
║    python -c "import nltk; nltk.download('vader_lexicon');                 ║
║               nltk.download('stopwords')"                                  ║
║                                                                             ║
║  DROP YOUR DATA:                                                            ║
║    data/minutes_pdfs/      ← MPC minutes PDFs from rbi.org.in             ║
║    data/gsec_daily.csv     ← 10Y G-Sec yield  (investing.com)             ║
║    data/walr_monthly.csv   ← WALR             (data.rbi.org.in)           ║
║    data/credit_monthly.csv ← Retail credit    (data.rbi.org.in)           ║
║    data/cpi_monthly.csv    ← CPI inflation    (mospi.gov.in / FRED)       ║
║                                                                             ║
║  THEN RUN:  python pipeline_final.py                                       ║
║  OUTPUTS:   outputs/figures/   outputs/tables/   outputs/results/          ║
╚══════════════════════════════════════════════════════════════════════════════╝

SECTION MAP
───────────
 §0  Imports and configuration
 §1  Rate decision database          ← hardcoded from RBI records
 §2  Data loading
     §2A  PDF text extraction
     §2B  Market data (G-Sec, INR/USD, WALR, credit, CPI)
 §3  ── NLP SECTION ──
     §3A  Text cleaning + TF-IDF
     §3B  VADER sentiment scoring
     §3C  Keyword frequency counts
 §4  ── QR + POWER ITERATION SECTION ──
     §4A  Gram-Schmidt QR decomposition    [Mathematics-I Item 6]
     §4B  Power iteration on AᵀA          [Mathematics-I Item 2]
     §4C  Hawkishness score construction
 §5  Master panel assembly
 §6  ── ML SECTION ──
     §6A  Logistic regression (statsmodels — p-values)
     §6B  Leave-One-Out cross-validation  (scikit-learn)
     §6C  Baseline comparisons
     §6D  Predicted probability time series
 §7  ── HYPOTHESIS TESTS ──
     §7A  H1 — Credibility (separating vs pooling equilibrium)
     §7B  H2 — Direct market response (G-Sec, INR/USD)
     §7C  H3 — Bank relay (WALR at lag 2-3)
     §7D  H4 — Retail exclusion
 §8  Figures
 §9  Export results
"""

# ══════════════════════════════════════════════════════════════════════════════
# §0  IMPORTS AND CONFIGURATION
# ══════════════════════════════════════════════════════════════════════════════

import os, re, glob, warnings, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from scipy import stats
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.metrics import (accuracy_score, confusion_matrix,
                              classification_report)
from sklearn.feature_extraction.text import TfidfVectorizer
import pdfplumber
import openpyxl

# Optional imports — graceful fallback if not installed
try:
    import statsmodels.api as sm
    HAS_STATSMODELS = True
except ImportError:
    HAS_STATSMODELS = False
    print("⚠ statsmodels not found — using numpy OLS fallback (install for p-values)")

try:
    import yfinance as yf
    HAS_YFINANCE = True
except ImportError:
    HAS_YFINANCE = False
    print("⚠ yfinance not found — download INR/USD manually")

try:
    import pandas_datareader as pdr
    HAS_PDR = True
except ImportError:
    HAS_PDR = False
    print("⚠ pandas_datareader not found — download G-Sec from FRED manually")

try:
    from nltk.sentiment.vader import SentimentIntensityAnalyzer
    from nltk.corpus import stopwords
    import nltk
    nltk.download('vader_lexicon', quiet=True)
    nltk.download('stopwords', quiet=True)
    STOPWORDS = set(stopwords.words('english'))
    vader = SentimentIntensityAnalyzer()
    HAS_NLTK = True
except Exception:
    HAS_NLTK = False
    STOPWORDS = {
        'the','a','an','and','or','but','in','on','at','to','for','of','with',
        'is','are','was','were','be','been','being','have','has','had','do',
        'does','did','will','would','could','should','may','might','shall',
        'this','that','these','those','it','its','by','from','as','into',
        'through','during','before','after','above','below','between',
    }
    print("⚠ nltk not found — using built-in sentiment word lists as fallback")

warnings.filterwarnings('ignore')
plt.rcParams.update({
    'figure.dpi': 150, 'font.family': 'serif',
    'axes.spines.top': False, 'axes.spines.right': False,
    'axes.grid': True, 'grid.alpha': 0.3,
})

BASE   = os.getcwd()
OUTFIG = os.path.join(BASE, 'outputs', 'figures')
OUTTAB = os.path.join(BASE, 'outputs', 'tables')
OUTRES = os.path.join(BASE, 'outputs', 'results')
for d in [OUTFIG, OUTTAB, OUTRES,
          os.path.join(BASE,'data','minutes_pdfs'),
          os.path.join(BASE,'data','minutes_text')]:
    os.makedirs(d, exist_ok=True)

# ══════════════════════════════════════════════════════════════════════════════
# §1  RATE DECISION DATABASE
#     Source: RBI press releases (rbi.org.in)
#     This is your TARGET VARIABLE for H1
# ══════════════════════════════════════════════════════════════════════════════

def load_rate_decisions():
    """
    All MPC decisions since inception (Oct 2016).
    Manually verified from RBI press releases.
    This is §1 — the anchor dataset everything is merged onto.
    """
    DECISIONS = [
        # (meeting_date, repo_rate_after, change_bps, decision)
        ("2016-10-04",6.25,-25,"cut"),  ("2016-12-07",6.25,0,"hold"),
        ("2017-02-08",6.25,0,"hold"),   ("2017-04-06",6.25,0,"hold"),
        ("2017-06-07",6.25,0,"hold"),   ("2017-08-02",6.00,-25,"cut"),
        ("2017-10-04",6.00,0,"hold"),   ("2017-12-06",6.00,0,"hold"),
        ("2018-02-07",6.00,0,"hold"),   ("2018-04-05",6.00,0,"hold"),
        ("2018-06-06",6.25,+25,"hike"), ("2018-08-01",6.50,+25,"hike"),
        ("2018-10-05",6.50,0,"hold"),   ("2018-12-05",6.50,0,"hold"),
        ("2019-02-07",6.25,-25,"cut"),  ("2019-04-04",6.00,-25,"cut"),
        ("2019-06-06",5.75,-25,"cut"),  ("2019-08-07",5.40,-35,"cut"),
        ("2019-10-04",5.15,-25,"cut"),  ("2019-12-05",5.15,0,"hold"),
        ("2020-02-06",5.15,0,"hold"),   ("2020-03-27",4.40,-75,"cut"),
        ("2020-05-22",4.00,-40,"cut"),  ("2020-08-06",4.00,0,"hold"),
        ("2020-10-09",4.00,0,"hold"),   ("2020-12-04",4.00,0,"hold"),
        ("2021-02-05",4.00,0,"hold"),   ("2021-04-07",4.00,0,"hold"),
        ("2021-06-04",4.00,0,"hold"),   ("2021-08-06",4.00,0,"hold"),
        ("2021-10-08",4.00,0,"hold"),   ("2021-12-08",4.00,0,"hold"),
        ("2022-02-10",4.00,0,"hold"),   ("2022-04-08",4.00,0,"hold"),
        ("2022-05-04",4.40,+40,"hike"), ("2022-06-08",4.90,+50,"hike"),
        ("2022-08-05",5.40,+50,"hike"), ("2022-09-30",5.90,+50,"hike"),
        ("2022-12-07",6.25,+35,"hike"), ("2023-02-08",6.50,+25,"hike"),
        ("2023-04-06",6.50,0,"hold"),   ("2023-06-08",6.50,0,"hold"),
        ("2023-08-10",6.50,0,"hold"),   ("2023-10-06",6.50,0,"hold"),
        ("2023-12-08",6.50,0,"hold"),   ("2024-02-08",6.50,0,"hold"),
        ("2024-04-05",6.50,0,"hold"),   ("2024-06-07",6.50,0,"hold"),
        ("2024-08-08",6.50,0,"hold"),   ("2024-10-09",6.50,0,"hold"),
        ("2024-12-06",6.50,0,"hold"),   ("2025-02-07",6.25,-25,"cut"),
        ("2025-04-09",6.00,-25,"cut"),  ("2025-06-06",5.75,-25,"cut"),
    ]
    df = pd.DataFrame(DECISIONS,
                      columns=['date','repo_rate','change_bps','decision'])
    df['date']   = pd.to_datetime(df['date'])
    df['hike']   = (df['decision']=='hike').astype(int)
    df['cut']    = (df['decision']=='cut').astype(int)
    df['change'] = (df['decision']!='hold').astype(int)
    return df.reset_index(drop=True)

# ══════════════════════════════════════════════════════════════════════════════
# §2A  DATA LOADING — PDF TEXT EXTRACTION
#      Input:  data/minutes_pdfs/*.pdf
#      Output: list of (date, text) per meeting
#      Used in: §3 NLP section
# ══════════════════════════════════════════════════════════════════════════════

def extract_text_from_pdf(path):
    """Extract and clean text from one PDF file using pdfplumber."""
    text = ''
    try:
        with pdfplumber.open(path) as pdf:
            for page in pdf.pages:
                t = page.extract_text()
                if t:
                    text += t + '\n'
    except Exception as e:
        print(f'    ⚠ PDF error [{os.path.basename(path)}]: {e}')
        return ''
    # Clean
    text = re.sub(r'\n{3,}', '\n\n', text)
    text = re.sub(r'[ \t]{2,}', ' ', text)
    text = re.sub(r'Page \d+ of \d+', '', text)
    text = re.sub(r'Reserve Bank of India\s*\n', '', text)
    return text.strip()


def load_corpus(decisions_df):
    """
    Load all PDFs from data/minutes_pdfs/.
    If none found, uses synthetic corpus for demonstration.
    Returns DataFrame with columns: date, text, source
    """
    pdf_dir = os.path.join(BASE, 'data', 'minutes_pdfs')
    txt_dir = os.path.join(BASE, 'data', 'minutes_text')
    pdfs = sorted(glob.glob(os.path.join(pdf_dir, '*.pdf')) +
                  glob.glob(os.path.join(pdf_dir, '*.PDF')))

    if not pdfs:
        print('  ⚠  No PDFs found → using synthetic corpus for demonstration')
        return _synthetic_corpus(decisions_df)

    print(f'  Found {len(pdfs)} PDFs')
    records = []
    for i, path in enumerate(pdfs):
        fname = os.path.basename(path)
        text  = extract_text_from_pdf(path)
        if not text:
            continue
        # Infer date from filename (e.g. 2022_06_08.pdf) or sequential order
        m = re.search(r'(\d{4})[-_](\d{2})[-_](\d{2})', fname)
        date = pd.to_datetime('-'.join(m.groups())) if m \
               else decisions_df['date'].iloc[min(i, len(decisions_df)-1)]
        # Save txt copy
        with open(os.path.join(txt_dir, fname.rsplit('.',1)[0]+'.txt'),
                  'w', encoding='utf-8') as f:
            f.write(text)
        records.append({'date': date, 'text': text, 'source': fname})
        print(f'    ✅ {fname[:50]:50s} {len(text.split()):5d} words')

    return pd.DataFrame(records)


def _synthetic_corpus(decisions_df):
    """Synthetic minutes for demonstration — replace with real PDFs."""
    HAWK = ('inflation elevated upside risk tighten vigilant withdrawal '
            'stance credibility repo increase persistent above target '
            'monetary tightening necessary price stability overshoot ')
    DOVE = ('growth accommodative recovery supportive stimulus below '
            'output gap fragile easing downside weak slowdown cut reduce '
            'demand pause hold neutral balanced maintain unchanged ')
    np.random.seed(42)
    records = []
    for _, row in decisions_df.iterrows():
        base  = HAWK if row['decision']=='hike' else \
                DOVE if row['decision']=='cut'  else (HAWK+DOVE)
        noise = ' '.join(np.random.choice(
            ['policy','committee','member','vote','rate','monetary',
             'economy','domestic','global','financial','market','outlook',
             'projection','assessment','review','quarter','growth','price',
             'inflation','forecast','data','conditions'], 40))
        records.append({
            'date': row['date'],
            'text': (base * 8) + noise,
            'source': 'synthetic'
        })
    return pd.DataFrame(records)

# ══════════════════════════════════════════════════════════════════════════════
# §2B  DATA LOADING — MARKET DATA
#      Inputs (drop CSVs in data/ folder):
#        gsec_daily.csv     — columns: date, gsec_yield
#        walr_monthly.csv   — columns: date, walr
#        credit_monthly.csv — columns: date, credit_growth
#        cpi_monthly.csv    — columns: date, cpi_yoy
#      Auto-downloads: INR/USD via yfinance, G-Sec via FRED (fallbacks)
#      Used in: §7 hypothesis tests (H2, H3, H4)
# ══════════════════════════════════════════════════════════════════════════════

def _load_csv_series(filename, col_rename):
    """Load a user-supplied CSV with date index."""
    path = os.path.join(BASE, 'data', filename)
    if not os.path.exists(path):
        return pd.DataFrame()
    df = pd.read_csv(path)
    df.columns = df.columns.str.strip().str.lower()
    date_col  = [c for c in df.columns if 'date' in c or 'month' in c or 'period' in c]
    value_col = [c for c in df.columns if c not in date_col]
    if not date_col or not value_col:
        print(f'  ⚠  {filename}: could not identify date/value columns')
        return pd.DataFrame()
    df = df[[date_col[0], value_col[0]]].copy()
    df.columns = ['date', col_rename]
    df['date'] = pd.to_datetime(df['date'], dayfirst=True, errors='coerce')
    df = df.dropna().set_index('date').sort_index()
    # Handle investing.com format where value might have commas
    if df[col_rename].dtype == object:
        df[col_rename] = df[col_rename].str.replace(',','').astype(float)
    return df


def load_market_data(decisions_df):
    """
    Load all market series and merge onto meeting dates.
    Returns the decisions_df with extra columns for each market variable.
    """
    panel = decisions_df.copy()

    # ── G-Sec 10Y Yield ──────────────────────────────────────────────────────
    gsec = _load_csv_series('gsec_daily.csv', 'gsec_yield')
    if gsec.empty and HAS_PDR:
        print('  🌐 Downloading G-Sec from FRED (monthly fallback)...')
        try:
            gsec = pdr.get_data_fred('INDIRLTLT01STM', start='2016-01-01')
            gsec.columns = ['gsec_yield']
        except Exception as e:
            print(f'  ⚠  FRED error: {e}')
    if not gsec.empty:
        panel = _merge_pre_post(panel, gsec, 'gsec_yield')
        print(f'  ✅ G-Sec yield: {len(gsec)} observations')

    # ── INR/USD ───────────────────────────────────────────────────────────────
    inrusd = _load_csv_series('inrusd_daily.csv', 'inrusd')
    if inrusd.empty and HAS_YFINANCE:
        print('  🌐 Downloading INR/USD from Yahoo Finance...')
        try:
            raw = yf.download('INR=X', start='2016-01-01', progress=False)
            col = ('Close','INR=X') if isinstance(raw.columns, pd.MultiIndex) \
                  else 'Close'
            inrusd = raw[col].to_frame('inrusd')
            inrusd.index = pd.to_datetime(inrusd.index)
        except Exception as e:
            print(f'  ⚠  yfinance error: {e}')
    if not inrusd.empty:
        panel = _merge_pre_post(panel, inrusd, 'inrusd')
        print(f'  ✅ INR/USD: {len(inrusd)} observations')

    # ── WALR (H3 — bank relay channel) ───────────────────────────────────────
    walr = _load_csv_series('walr_monthly.csv', 'walr')
    if not walr.empty:
        panel = _merge_pre_post(panel, walr, 'walr')
        print(f'  ✅ WALR: {len(walr)} observations')
    else:
        print('  ⚠  walr_monthly.csv missing → H3 will be skipped')
        print('     Get from: data.rbi.org.in → Interest Rates → WALR')

    # ── Retail Credit Growth (H4 — retail exclusion) ─────────────────────────
    credit = _load_csv_series('credit_monthly.csv', 'credit_growth')
    if not credit.empty:
        panel = _merge_pre_post(panel, credit, 'credit_growth')
        print(f'  ✅ Retail credit growth: {len(credit)} observations')
    else:
        print('  ⚠  credit_monthly.csv missing → H4 will be skipped')
        print('     Get from: data.rbi.org.in → Banking → Sectoral Deployment')

    # ── CPI (control variable) ────────────────────────────────────────────────
    cpi = _load_csv_series('cpi_monthly.csv', 'cpi_yoy')
    if cpi.empty and HAS_PDR:
        try:
            raw = pdr.get_data_fred('INDCPIALLMINMEI', start='2016-01-01')
            raw.columns = ['cpi_level']
            cpi = raw['cpi_level'].pct_change(12).mul(100).to_frame('cpi_yoy').dropna()
        except Exception:
            pass
    if not cpi.empty:
        panel = _merge_pre_post(panel, cpi, 'cpi_yoy')
        print(f'  ✅ CPI: {len(cpi)} observations')

    return panel


def _merge_pre_post(panel, series, col):
    """
    For each MPC meeting date, find the value immediately BEFORE
    and immediately AFTER in the series.
    Computes pre, post, and change columns for event-study analysis.
    """
    series = series.copy()
    series.index = pd.to_datetime(series.index)
    pre_vals, post_vals = [], []
    for d in panel['date']:
        pre  = series[series.index <= d]
        post = series[series.index >  d]
        pre_vals.append(pre[col].iloc[-1]  if not pre.empty  else np.nan)
        post_vals.append(post[col].iloc[0] if not post.empty else np.nan)
    panel = panel.copy()
    panel[f'{col}_pre']    = pre_vals
    panel[f'{col}_post']   = post_vals
    panel[f'{col}_change'] = np.array(post_vals) - np.array(pre_vals)
    return panel

# ══════════════════════════════════════════════════════════════════════════════
# §3  NLP SECTION
#     §3A  Text cleaning and TF-IDF
#     §3B  VADER sentiment scoring
#     §3C  Keyword frequency counts
#
#     PURPOSE: Convert raw text into numerical features.
#     These are INPUTS to the QR/eigenvector computation in §4.
# ══════════════════════════════════════════════════════════════════════════════

# §3A — Text cleaning
def clean_text(text):
    """
    NLP §3A: Lowercase, remove punctuation/numbers,
    remove stopwords. Prepares text for TF-IDF.
    """
    text = text.lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    tokens = [w for w in text.split()
              if w not in STOPWORDS and len(w) > 2]
    return ' '.join(tokens)


# §3B — Sentiment scoring
# Built-in financial sentiment word lists (fallback when NLTK unavailable)
# Based on Loughran & McDonald (2011) — the standard in financial NLP
_POSITIVE_WORDS = {
    'recovery','growth','strong','robust','positive','improve','stable',
    'resilient','momentum','confidence','supported','favorable','healthy',
    'buoyant','uptick','optimistic','balanced','contained','anchored',
}
_NEGATIVE_WORDS = {
    'risk','concern','elevated','pressures','uncertainty','volatile',
    'deteriorate','weak','slowdown','adverse','fragile','stress','tight',
    'inflationary','persistent','upside','vigilant','cautious','headwinds',
}

def compute_sentiment(text):
    """
    NLP §3B: Returns a sentiment score in [-1, +1].
    Uses VADER if available (better), else Loughran-McDonald word lists.
    Note: in CB text, negative words often signal hawkishness (concern
    about inflation), so this is a cross-check, not the main score.
    """
    if HAS_NLTK:
        return vader.polarity_scores(text)['compound']
    words = re.findall(r'\b[a-z]+\b', text.lower())
    if not words:
        return 0.0
    pos = sum(1 for w in words if w in _POSITIVE_WORDS)
    neg = sum(1 for w in words if w in _NEGATIVE_WORDS)
    total = pos + neg
    return (pos - neg) / total if total > 0 else 0.0


# §3C — Keyword frequency (domain-specific features)
HAWK_WORDS = {
    'inflation','inflationary','upside','tighten','tightening','vigilant',
    'withdrawal','elevated','persistent','above','target','concern',
    'credibility','increase','stance','focused','hike','tight','restrictive',
    'overshoot','price','stability','risk','basis','points',
}
DOVE_WORDS = {
    'accommodative','supportive','growth','recovery','stimulus','below',
    'slack','gap','fragile','easing','cut','reduce','support','demand',
    'downside','weak','slowdown','maintain','unchanged','pause','hold',
    'neutral','balanced','space',
}

def keyword_score(text):
    """
    NLP §3C: Simple dictionary-based hawkishness score.
    (hawk_count - dove_count) / total_policy_words
    Used as CROSS-CHECK against LSA score from §4.
    This is what the code you showed earlier was doing implicitly.
    """
    words = re.findall(r'\b[a-z]+\b', text.lower())
    h = sum(1 for w in words if w in HAWK_WORDS)
    d = sum(1 for w in words if w in DOVE_WORDS)
    return (h - d) / (h + d) if (h + d) > 0 else 0.0


def compute_keyword_frequencies(text):
    """NLP §3C: Raw counts of hawkish and dovish words."""
    words = re.findall(r'\b[a-z]+\b', text.lower())
    hawk_count = sum(1 for w in words if w in HAWK_WORDS)
    dove_count = sum(1 for w in words if w in DOVE_WORDS)
    inflation_count = sum(1 for w in words if w in
                          {'inflation','inflationary','cpi','price','prices'})
    growth_count    = sum(1 for w in words if w in
                          {'growth','gdp','output','recovery','expansion'})
    return hawk_count, dove_count, inflation_count, growth_count

# ══════════════════════════════════════════════════════════════════════════════
# §4  QR DECOMPOSITION + POWER ITERATION — HAWKISHNESS SCORE
#     This is the MATHEMATICAL CORE of the project.
#     Implements Mathematics-I Items 6 and 2 from scratch.
#
#     WHAT THIS SECTION DOES:
#       Input:  TF-IDF matrix A  (n_docs × vocab_size)
#       Step 1: A = QR via Modified Gram-Schmidt  → orthonormal basis Q
#       Step 2: Power iteration on AᵀA = RᵀR     → dominant eigenvectors V
#       Step 3: Project documents: scores = A @ V  → one score per meeting
#       Step 4: Identify hawkishness axis by word loading inspection
#       Output: hₜ — one standardised number per meeting
# ══════════════════════════════════════════════════════════════════════════════

# §4A — Gram-Schmidt QR Decomposition  [Mathematics-I Item 6]
def gram_schmidt_qr(A, max_cols=200):
    """
    QR DECOMPOSITION via Modified Gram-Schmidt algorithm.
    ─────────────────────────────────────────────────────
    Mathematics-I Item 6 (Modified Gram-Schmidt / Householder).

    Given matrix A (m × n), finds Q (m × n) and R (n × n) such that:
        A = Q @ R
    where Q has orthonormal columns (Qᵀ Q = I) and R is upper triangular.

    WHY THIS MATTERS HERE:
    - Q gives an orthonormal basis for the document space
    - This removes correlations between words (e.g. 'inflation' and
      'inflationary' are correlated — QR orthogonalises them)
    - The decomposition ensures AᵀA = RᵀRᵀQᵀQR = RᵀR, which is
      why power iteration on AᵀA is numerically well-conditioned

    Modified Gram-Schmidt (vs classical) is numerically stable:
    it re-orthogonalises at each step rather than using the original vectors.

    Args:
        A:        TF-IDF matrix (n_docs × vocab_size)
        max_cols: limit for computational feasibility
    Returns:
        Q: orthonormal basis  (n_docs × max_cols)
        R: upper triangular   (max_cols × max_cols)
    """
    A_ = A[:, :max_cols].copy().astype(float)
    m, n = A_.shape
    Q = np.zeros((m, n))
    R = np.zeros((n, n))

    for j in range(n):
        # Take the j-th column of A
        v = A_[:, j].copy()

        # Orthogonalise against all previous Q columns
        # (Modified GS: subtract projections one by one, not all at once)
        for i in range(j):
            R[i, j] = np.dot(Q[:, i], A_[:, j])  # projection coefficient
            v       -= R[i, j] * Q[:, i]           # subtract projection

        # Normalise to get unit vector
        R[j, j] = np.linalg.norm(v)
        if R[j, j] > 1e-10:
            Q[:, j] = v / R[j, j]
        # else: linearly dependent column — leave as zero

    return Q, R


# §4B — Power Iteration  [Mathematics-I Item 2]
def power_iteration(M, num_vecs=5, n_iter=150, seed=42):
    """
    POWER ITERATION for dominant eigenvectors of symmetric matrix M.
    ────────────────────────────────────────────────────────────────
    Mathematics-I Item 2 (Eigenvalue Computation — Power Iteration).

    For our application: M = AᵀA  (vocab × vocab covariance matrix)
    The dominant eigenvectors of AᵀA are the right singular vectors of A,
    which define the latent topic directions in the document space.

    Algorithm (deflation method to find multiple eigenvectors):
    1. Start with a random vector b
    2. Repeatedly multiply: b ← M @ b, then normalise
    3. Converges to the eigenvector with largest eigenvalue
    4. Deflate: M ← M - λ vvᵀ  (remove found eigenvector's contribution)
    5. Repeat for next eigenvector

    CONNECTION TO QR:
    The QR algorithm for eigenvalues is: Aₖ₊₁ = Rₖ Qₖ  (repeated QR decomp)
    Power iteration is a simplified, single-vector version of this.
    Both exploit the same mathematical property: repeated multiplication
    amplifies the dominant eigenvector direction.

    Args:
        M:        symmetric matrix (AᵀA), shape (vocab × vocab)
        num_vecs: number of eigenvectors to extract
        n_iter:   iterations per eigenvector (more = more accurate)
    Returns:
        V: matrix of eigenvectors, shape (vocab × num_vecs)
           Each column is a direction in word-space (a latent topic)
    """
    rng   = np.random.default_rng(seed)
    V     = []
    M_def = M.copy().astype(float)

    for k in range(num_vecs):
        # Initialise with random unit vector
        b = rng.standard_normal(M_def.shape[1])
        b /= np.linalg.norm(b)

        # Power iteration: b converges to dominant eigenvector of M_def
        for _ in range(n_iter):
            b_new = M_def @ b
            norm  = np.linalg.norm(b_new)
            if norm < 1e-12:
                break
            b = b_new / norm

        # Corresponding eigenvalue: λ = bᵀ M b
        eigenval = float(b @ M_def @ b)
        V.append(b)

        # Deflation: remove this eigenvector's contribution from M
        # So next iteration finds the NEXT largest eigenvector
        M_def = M_def - eigenval * np.outer(b, b)

    return np.column_stack(V)  # shape: (vocab, num_vecs)


# §4C — Build hawkishness scores
def build_hawkishness_scores(corpus_df, n_components=5, vocab_size=2000):
    """
    FULL LSA PIPELINE:  Text → TF-IDF → QR → Power Iteration → hₜ
    ─────────────────────────────────────────────────────────────────
    Step 1 (NLP §3A):   Clean text, build TF-IDF matrix A
    Step 2 (QR  §4A):   Decompose A = QR (Gram-Schmidt)
    Step 3 (PI  §4B):   Power iteration on AᵀA to get eigenvectors V
    Step 4 (NLP §4C):   Project docs: scores = A @ V
    Step 5 (NLP §4C):   Identify hawkishness axis by word loadings
    Step 6 (NLP §4C):   Standardise → output hₜ

    Returns: corpus_df with hawk_score, sentiment, keyword features added
             V (eigenvectors), vocab array, best_k (hawkishness component index)
    """
    print('\n  ── §3A: TF-IDF vectorisation ──')
    cleaned = corpus_df['text'].apply(clean_text)
    vect = TfidfVectorizer(
        max_features=vocab_size,
        min_df=2, max_df=0.95,
        ngram_range=(1, 2),
        sublinear_tf=True,
    )
    A     = vect.fit_transform(cleaned).toarray()  # shape: (n_docs, vocab)
    vocab = np.array(vect.get_feature_names_out())
    print(f'     Term-document matrix A: {A.shape[0]} documents × {A.shape[1]} terms')

    print('\n  ── §4A: QR decomposition (Modified Gram-Schmidt) ──')
    # We decompose A using a subset of columns for efficiency
    # Full decomposition of large sparse TF-IDF matrix is expensive
    max_cols = min(A.shape[0], 150)  # n_docs is the binding constraint
    Q, R     = gram_schmidt_qr(A, max_cols=max_cols)
    print(f'     Q: {Q.shape}  (orthonormal basis for document space)')
    print(f'     R: {R.shape}  (upper triangular)')
    print(f'     Verification: ||QᵀQ - I||_F = '
          f'{np.linalg.norm(Q.T @ Q - np.eye(Q.shape[1])):.2e}  (should be ~0)')

    print('\n  ── §4B: Power iteration on AᵀA ──')
    # AᵀA = RᵀRᵀQᵀQR = RᵀR  (by QR decomposition, since QᵀQ = I)
    # This is why QR makes the eigenvalue computation cleaner
    ATA = A.T @ A  # vocab × vocab covariance matrix
    V   = power_iteration(ATA, num_vecs=n_components, n_iter=150)
    print(f'     V: {V.shape}  ({n_components} dominant eigenvectors of AᵀA)')
    print(f'     Each column = one latent topic direction in word-space')

    print('\n  ── §4C: Projecting documents → topic scores ──')
    # Each row = one meeting, each column = coordinate on one topic axis
    topic_scores = A @ V  # shape: (n_docs, n_components)
    print(f'     Topic score matrix: {topic_scores.shape}')

    print('\n  ── §4C: Identifying hawkishness axis ──')
    # For each eigenvector, compute how well its top words align with
    # the known hawkish/dovish vocabulary
    # The machine found these directions — we now LABEL them
    alignment_scores = []
    for k in range(n_components):
        loadings = V[:, k]
        top_pos  = set(vocab[np.argsort(loadings)[-30:]])   # high-loading words
        top_neg  = set(vocab[np.argsort(loadings)[:30]])    # low-loading words
        # Positive alignment: top words are hawkish, bottom words are dovish
        score = (len(top_pos & HAWK_WORDS) - len(top_pos & DOVE_WORDS) +
                 len(top_neg & DOVE_WORDS) - len(top_neg & HAWK_WORDS))
        alignment_scores.append((k, score))
        print(f'     Component {k+1}: hawkishness alignment = {score:+d}')

    best_k = max(alignment_scores, key=lambda x: x[1])[0]
    print(f'     → Component {best_k+1} selected as hawkishness axis')

    # Report what words define the axis
    loadings      = V[:, best_k]
    hawk_end      = vocab[np.argsort(loadings)[-12:]][::-1]
    dove_end      = vocab[np.argsort(loadings)[:12]]
    print(f'\n     Hawkish end: {", ".join(hawk_end)}')
    print(f'     Dovish end:  {", ".join(dove_end)}')

    print('\n  ── §4C: Computing final scores ──')
    raw_scores = topic_scores[:, best_k]

    # Build output dataframe
    out = corpus_df.copy().sort_values('date').reset_index(drop=True)
    out['hawk_score_raw'] = raw_scores

    # Standardise: mean 0, std 1
    mu, sigma    = raw_scores.mean(), raw_scores.std()
    out['hawk_score'] = (raw_scores - mu) / (sigma if sigma > 1e-9 else 1)

    # §3B — Sentiment scores (NLP feature, separate from LSA)
    out['sentiment'] = out['text'].apply(compute_sentiment)

    # §3C — Keyword features (NLP feature, cross-check for LSA)
    kw = out['text'].apply(
        lambda t: pd.Series(compute_keyword_frequencies(t),
                            index=['hawk_freq','dove_freq',
                                   'inflation_freq','growth_freq']))
    out = pd.concat([out, kw], axis=1)
    out['dict_score'] = out['text'].apply(keyword_score)

    # Lag features (avoids look-ahead bias — critical for H1)
    out['hawk_lag1']   = out['hawk_score'].shift(1)
    out['hawk_lag2']   = out['hawk_score'].shift(2)
    out['hawk_lag3']   = out['hawk_score'].shift(3)
    out['hawk_change'] = out['hawk_score'].diff()   # momentum
    out['hawk_ma3']    = out['hawk_score'].rolling(3, min_periods=1).mean()

    return out, V, vocab, best_k

# ══════════════════════════════════════════════════════════════════════════════
# §5  MASTER PANEL ASSEMBLY
#     Merge NLP scores from §3-4 onto market data from §2B
#     Output: one row per MPC meeting, all variables aligned
# ══════════════════════════════════════════════════════════════════════════════

def build_master_panel(nlp_df, market_df):
    """
    Merge NLP/LSA scores onto the market data panel.
    Each meeting gets its NLP score from the minutes document
    dated at or before that meeting date.
    """
    panel    = market_df.copy()
    panel['date'] = pd.to_datetime(panel['date'])
    nlp_df   = nlp_df.sort_values('date')

    nlp_cols = ['hawk_score','hawk_lag1','hawk_lag2','hawk_lag3',
                'hawk_change','hawk_ma3','dict_score',
                'sentiment','hawk_freq','dove_freq',
                'inflation_freq','growth_freq']

    for col in nlp_cols:
        if col not in nlp_df.columns:
            continue
        vals = []
        for d in panel['date']:
            pre = nlp_df[nlp_df['date'] <= d]
            vals.append(pre[col].iloc[-1] if not pre.empty else np.nan)
        panel[col] = vals

    panel = panel.dropna(subset=['hawk_score']).reset_index(drop=True)

    # Covid regime dummy — controls for pandemic language dominating LSA variance
    panel['covid_dummy'] = (
        (panel['date'] >= '2020-03-01') &
        (panel['date'] <= '2021-12-31')
    ).astype(int)
    print(f'  Covid dummy: {panel["covid_dummy"].sum()} meetings flagged (Mar 2020 – Dec 2021)')

    # Flip sign if needed: higher hawk_score should precede hikes
    if panel['hike'].sum() > 0:
        mean_hike    = panel[panel['hike']==1]['hawk_score'].mean()
        mean_nonhike = panel[panel['hike']==0]['hawk_score'].mean()
        if mean_hike < mean_nonhike:
            print('  ↺ Flipping sign: making positive = hawkish')
            for c in ['hawk_score','hawk_lag1','hawk_lag2','hawk_lag3',
                      'hawk_change','hawk_ma3']:
                if c in panel.columns:
                    panel[c] = -panel[c]

    print(f'  Master panel: {len(panel)} meetings × {len(panel.columns)} variables')
    return panel

# ══════════════════════════════════════════════════════════════════════════════
# §6  ML SECTION — CLASSIFICATION
#     §6A  Logistic regression with p-values (statsmodels)
#     §6B  Leave-One-Out cross-validation (scikit-learn)
#     §6C  Baseline comparisons
#     §6D  Predicted probabilities
#
#     PURPOSE: Predict rate hike vs non-hike from hawkishness score.
#     This is the empirical test of H1 (credibility).
#     The ML output — P(hike) — also becomes input to the signaling test.
# ══════════════════════════════════════════════════════════════════════════════

def _ols_numpy(X, y):
    """
    OLS via numpy (fallback when statsmodels unavailable).
    Returns coefficients and approximate standard errors.
    """
    XtX_inv = np.linalg.pinv(X.T @ X)
    beta    = XtX_inv @ X.T @ y
    resid   = y - X @ beta
    s2      = np.dot(resid, resid) / (len(y) - X.shape[1])
    se      = np.sqrt(s2 * np.diag(XtX_inv))
    t_vals  = beta / (se + 1e-12)
    p_vals  = 2 * (1 - stats.t.cdf(np.abs(t_vals), df=len(y)-X.shape[1]))
    return beta, se, t_vals, p_vals


def run_ml_section(panel):
    """
    §6 ML SECTION — complete classification pipeline.
    Returns dict of results including LOO accuracy, coefficients, probabilities.
    """
    print('\n' + '═'*60)
    print('§6 ML SECTION — LOGISTIC REGRESSION + LOO-CV')
    print('═'*60)

    # Feature selection — use what's available
    candidate_feats = ['hawk_lag1','hawk_change','sentiment',
                       'inflation_freq','dict_score','cpi_yoy','covid_dummy']
    feats = [f for f in candidate_feats if f in panel.columns]
    sub   = panel[feats + ['hike','date','decision']].dropna()

    if len(sub) < 10:
        print('  ⚠  Insufficient data for ML section')
        return {}

    X      = sub[feats].values.astype(float)
    y      = sub['hike'].values
    scaler = StandardScaler()
    X_sc   = scaler.fit_transform(X)

    print(f'\n  Features: {feats}')
    print(f'  Observations: {len(sub)}  |  Hikes: {y.sum()}  |  Non-hikes: {(1-y).sum()}')

    # §6A — Logistic regression with p-values ─────────────────────────────────
    print('\n  ── §6A: Logistic regression (for paper coefficients) ──')
    if HAS_STATSMODELS:
        X_sm = sm.add_constant(X_sc)
        try:
            logit  = sm.Logit(y, X_sm).fit(disp=False)
            print(logit.summary())
            coef_df = pd.DataFrame({
                'feature': ['const'] + feats,
                'coef':    logit.params,
                'se':      logit.bse,
                'p_value': logit.pvalues,
            })
        except Exception as e:
            print(f'  Logit error: {e}')
            coef_df = pd.DataFrame()
    else:
        # Numpy fallback: linear probability model (OLS on binary y)
        print('  (statsmodels unavailable — using linear probability model)')
        X_sm     = np.column_stack([np.ones(len(X_sc)), X_sc])
        b, se, t, p = _ols_numpy(X_sm, y.astype(float))
        coef_df  = pd.DataFrame({
            'feature': ['const'] + feats,
            'coef': b, 'se': se, 't_stat': t, 'p_value': p,
        })
        print(f'\n  {"Feature":<22} {"Coef":>8} {"SE":>8} {"p-val":>8}')
        print('  ' + '-'*50)
        for _, row in coef_df.iterrows():
            sig = '***' if row['p_value']<0.01 else '**' if row['p_value']<0.05 \
                  else '*' if row['p_value']<0.10 else ''
            print(f'  {row["feature"]:<22} {row["coef"]:>+8.3f} '
                  f'{row["se"]:>8.3f} {row["p_value"]:>8.4f} {sig}')

    # §6B — Leave-One-Out cross-validation ────────────────────────────────────
    print('\n  ── §6B: Leave-One-Out cross-validation ──')
    clf   = LogisticRegression(C=0.5, max_iter=500, random_state=42)
    loo   = LeaveOneOut()
    y_hat = cross_val_predict(clf, X_sc, y, cv=loo)
    acc   = accuracy_score(y, y_hat)

    print(f'  LOO-CV accuracy:         {acc:.1%}')
    print(f'\n{classification_report(y, y_hat, target_names=["Non-hike","Hike"])}')
    print(f'\n  Confusion matrix:')
    print(f'  {confusion_matrix(y, y_hat)}')

    # §6C — Baseline comparisons ──────────────────────────────────────────────
    print('\n  ── §6C: Baseline comparisons ──')
    maj_base  = max(y.mean(), 1 - y.mean())
    pers_base = accuracy_score(y, pd.Series(y).shift(1).fillna(0).astype(int))
    lift      = acc - maj_base

    print(f'  Majority-class baseline: {maj_base:.1%}')
    print(f'  Persistence baseline:    {pers_base:.1%}')
    print(f'  Your model:              {acc:.1%}')
    print(f'  Lift over majority:      {lift:+.1%}')

    if lift > 0.05:
        print('  ✅ Model beats baseline — hawkishness score has predictive power')
    else:
        print('  ⚠  Modest lift — consider whether data is sufficient')

    # §6D — Predicted probabilities ───────────────────────────────────────────
    print('\n  ── §6D: Predicted probabilities ──')
    clf.fit(X_sc, y)
    probs      = clf.predict_proba(X_sc)[:, 1]
    sub        = sub.copy()
    sub['p_hike'] = probs

    return {
        'features': feats, 'X_sc': X_sc, 'y': y,
        'clf': clf, 'scaler': scaler,
        'y_hat': y_hat, 'probs': probs,
        'acc': acc, 'maj_base': maj_base,
        'coef_df': coef_df if 'coef_df' in dir() else pd.DataFrame(),
        'sub': sub,
    }

# ══════════════════════════════════════════════════════════════════════════════
# §7  HYPOTHESIS TESTS
#     Direct implementation of H1–H4 from the theoretical section.
#     Each test maps exactly to one equation in the theory section.
# ══════════════════════════════════════════════════════════════════════════════

def h1_credibility(panel):
    """
    H1 — CREDIBILITY TEST
    ─────────────────────
    Theory:   Separating vs pooling equilibrium (Crawford-Sobel 1982)
    Test:     Does hawk_lag1 predict the rate decision?
    Equation: hike_t = α + β·h_{t-1} + ε
    Result:   β > 0 and significant → separating eq → credible signal
              β insignificant       → pooling eq    → cheap talk

    Uses: t-test of means + logistic regression coefficient
    """
    print('\n' + '═'*60)
    print('H1 — CREDIBILITY (Separating vs Pooling Equilibrium)')
    print('═'*60)

    res = {}

    # T-test of means (simplest credibility test)
    hike_s    = panel[panel['hike']==1]['hawk_lag1'].dropna()
    nonhike_s = panel[panel['hike']==0]['hawk_lag1'].dropna()

    if len(hike_s) > 1 and len(nonhike_s) > 1:
        t, p = stats.ttest_ind(hike_s, nonhike_s, equal_var=False)
        res  = {'t': t, 'p': p,
                'mean_hike': hike_s.mean(),
                'mean_nonhike': nonhike_s.mean()}
        print(f'\n  Mean hawk score pre-hike:     {hike_s.mean():+.3f}')
        print(f'  Mean hawk score pre-non-hike: {nonhike_s.mean():+.3f}')
        print(f'  Difference:                   {hike_s.mean()-nonhike_s.mean():+.3f}')
        print(f'  Welch t-test: t = {t:.3f},  p = {p:.4f}')
        print(f'\n  Verdict: ', end='')
        if p < 0.05:
            print('✅ REJECT H0 → Separating equilibrium → Credible signal')
        elif p < 0.10:
            print('⚠  WEAK EVIDENCE → Marginal credibility (p < 0.10)')
        else:
            print('❌ FAIL TO REJECT H0 → Pooling equilibrium → Cheap talk')
    else:
        print('  ⚠  Insufficient hike observations for t-test')

    # Ex-Covid robustness test
    if 'covid_dummy' in panel.columns:
        print('[Excluding Covid 2020-2021]')
        panel_ex   = panel[panel['covid_dummy']==0]
        hike_ex    = panel_ex[panel_ex['hike']==1]['hawk_lag1'].dropna()
        nonhike_ex = panel_ex[panel_ex['hike']==0]['hawk_lag1'].dropna()
        if len(hike_ex) > 1 and len(nonhike_ex) > 1:
            t_ex, p_ex = stats.ttest_ind(hike_ex, nonhike_ex, equal_var=False)
            print(f'  Meetings kept: {len(panel_ex)}  Hikes: {len(hike_ex)}  Non-hikes: {len(nonhike_ex)}')
            print(f'  Mean pre-hike:     {hike_ex.mean():+.3f}')
            print(f'  Mean pre-non-hike: {nonhike_ex.mean():+.3f}')
            print(f'  t = {t_ex:.3f},  p = {p_ex:.4f}')
            if p_ex < 0.05:
                print('  SIGNIFICANT ex-Covid -> Covid was masking the signal')
            elif p_ex < 0.10:
                print('  MARGINAL ex-Covid -> partial improvement')
            else:
                print('  Still insignificant -> LSA axis needs recalibration')

    return res


def h2_market_response(panel):
    """
    H2 — DIRECT MARKET RESPONSE
    ────────────────────────────
    Theory:   Institutional investors (bond markets, FX) observe signal directly
    Test:     Does h_t predict G-Sec yield change and INR/USD change?
    Equation: ΔYield_{t+1} = α + β·h_t + ε     [eq 3.4 in theory section]
    Result:   β significant → markets respond to communication before decision
              β insignificant → markets wait for actual rate announcement

    Agent type: Institutional investors (θᵢ closest to median)
    """
    print('\n' + '═'*60)
    print('H2 — DIRECT MARKET RESPONSE (Institutional Agents)')
    print('═'*60)

    targets = {
        'gsec_yield_change': '10Y G-Sec Yield Change (pp)',
        'inrusd_change':     'INR/USD Change',
    }
    res = {}

    for col, label in targets.items():
        if col not in panel.columns:
            print(f'\n  ⚠  {label}: data not available')
            print(f'     Add data/gsec_daily.csv or data/inrusd_daily.csv')
            continue

        feats = [f for f in ['hawk_lag1','hawk_change','change_bps','covid_dummy']
                 if f in panel.columns]
        sub   = panel[feats + [col]].dropna()

        if len(sub) < 8:
            print(f'\n  ⚠  {label}: only {len(sub)} obs — skipping')
            continue

        X = np.column_stack([np.ones(len(sub))] +
                            [sub[f].values for f in feats])
        y = sub[col].values.astype(float)

        b, se, t_v, p_v = _ols_numpy(X, y)
        r2 = 1 - np.var(y - X @ b) / np.var(y)

        res[col] = {'coef': b[1], 'p': p_v[1], 'r2': r2}
        sig = '***' if p_v[1]<0.01 else '**' if p_v[1]<0.05 \
              else '*' if p_v[1]<0.10 else '(ns)'

        print(f'\n  {label}')
        print(f'    hawk coef = {b[1]:+.4f}  p = {p_v[1]:.4f} {sig}  R² = {r2:.3f}')
        print(f'    Verdict: ', end='')
        if p_v[1] < 0.10:
            print('✅ Signal reaches institutional agents directly')
        else:
            print('⚠  No significant direct market response')

    return res


def h3_bank_relay(panel):
    """
    H3 — BANK RELAY (Intermediation Channel)
    ─────────────────────────────────────────
    Theory:   Banks transmit signal to retail via lending rates, lag τ=2-3
    Test:     Does h_t predict WALR change at lag 2-3?
    Equation: ΔWALR_{t+2} = α + β·h_t + γ·r_t + ε   [eq 3.5 in theory section]
    Result:   β significant (controlling for r_t) → banks respond to communication
              not just the rate decision itself

    Agent type: Commercial banks (intermediate ideal point)
    Data needed: data/walr_monthly.csv from RBI DBIE
    """
    print('\n' + '═'*60)
    print('H3 — BANK RELAY (Bank Intermediation Channel)')
    print('═'*60)

    walr_col = next((c for c in panel.columns if c == 'walr_change'), None)
    if walr_col is None:
        print('  ⚠  WALR data not available — H3 skipped')
        print('     Download: data.rbi.org.in → Interest Rates → WALR')
        print('     Save as:  data/walr_monthly.csv  (columns: date, walr)')
        return {}

    res = {}
    p   = panel.copy().sort_values('date').reset_index(drop=True)

    for lag in [1, 2, 3]:
        p[f'hawk_l{lag}'] = p['hawk_score'].shift(lag)
        ctrl = [f'hawk_l{lag}', 'change_bps'] + (['covid_dummy'] if 'covid_dummy' in p.columns else [])
        sub  = p[ctrl + [walr_col]].dropna()
        if len(sub) < 8:
            continue

        X    = np.column_stack([np.ones(len(sub))] + [sub[c].values for c in ctrl])
        y    = sub[walr_col].values.astype(float)
        b, se, t_v, p_v = _ols_numpy(X, y)
        r2   = 1 - np.var(y - X @ b) / np.var(y)
        sig  = '***' if p_v[1]<0.01 else '**' if p_v[1]<0.05 \
               else '*' if p_v[1]<0.10 else '(ns)'

        res[f'lag{lag}'] = {'coef': b[1], 'p': p_v[1], 'r2': r2}
        print(f'  Lag {lag}: β(hawk) = {b[1]:+.4f}  p = {p_v[1]:.4f} {sig}  '
              f'β(rate_decision) = {b[2]:+.4f}  R² = {r2:.3f}')

    if res:
        best = min(res, key=lambda k: res[k]['p'])
        print(f'\n  Best lag: {best}  (p = {res[best]["p"]:.4f})')
        if res[best]['p'] < 0.10:
            print('  ✅ Bank relay confirmed — WALR responds to RBI communication')
            print('     (even after controlling for actual rate decision)')
        else:
            print('  ⚠  Weak/no bank relay evidence')

    return res


def h4_retail_exclusion(panel):
    """
    H4 — RETAIL EXCLUSION
    ──────────────────────
    Theory:   Retail agents (low θᵢ) don't respond to signal directly;
              they respond only after WALR changes (bank relay)
    Test:     Two models compared:
              Model 1: credit_growth = α + β·h_t + ε             (direct)
              Model 2: credit_growth = α + β·h_t + δ·ΔWALR + ε  (mediated)
              H4 confirmed if: β insignificant in Model 1
                            OR β significant in M1 but drops in M2
    Equation: Retail credit_{t+3} = α'' + δ·WALR_{t+2} + ε    [eq 3.6]

    Agent type: Retail borrowers (lowest θᵢ — prefer low rates)
    Data needed: data/credit_monthly.csv from RBI DBIE
    """
    print('\n' + '═'*60)
    print('H4 — RETAIL EXCLUSION (Single-Peaked Preference Result)')
    print('═'*60)

    credit_col = next((c for c in panel.columns if c == 'credit_growth_change'), None)
    walr_col   = next((c for c in panel.columns if c == 'walr_change'), None)

    if credit_col is None:
        print('  ⚠  Retail credit data not available — H4 skipped')
        print('     Download: data.rbi.org.in → Banking → Sectoral Deployment')
        print('     Save as:  data/credit_monthly.csv  (columns: date, credit_growth)')
        return {}

    base_feats = [f for f in ['hawk_lag1','change_bps','cpi_yoy','covid_dummy']
                  if f in panel.columns]
    sub = panel[base_feats + [credit_col] +
                ([walr_col] if walr_col else [])].dropna()

    if len(sub) < 8:
        print(f'  ⚠  Only {len(sub)} obs after dropna — skipping')
        return {}

    y = sub[credit_col].values.astype(float)

    # Model 1 — direct effect of hawk_score on retail credit
    X1     = np.column_stack([np.ones(len(sub))] +
                              [sub[f].values for f in base_feats])
    b1, _, _, p1 = _ols_numpy(X1, y)
    r2_m1  = 1 - np.var(y - X1 @ b1) / np.var(y)

    print(f'\n  Model 1 (direct):  hawk_lag1 p = {p1[1]:.4f}  R² = {r2_m1:.3f}')

    res = {'direct_p': p1[1]}

    # Model 2 — mediated through WALR
    if walr_col:
        walr_feats = base_feats + [walr_col]
        X2     = np.column_stack([np.ones(len(sub))] +
                                  [sub[f].values for f in walr_feats])
        b2, _, _, p2 = _ols_numpy(X2, y)
        r2_m2  = 1 - np.var(y - X2 @ b2) / np.var(y)
        print(f'  Model 2 (+ WALR): hawk_lag1 p = {p2[1]:.4f}  R² = {r2_m2:.3f}')
        res['mediated_p'] = p2[1]

        print('\n  Verdict: ', end='')
        if p1[1] >= 0.10:
            print('✅ H4 confirmed: no direct effect — retail excluded from signal')
        elif p1[1] < 0.10 and p2[1] >= 0.10:
            print('✅ H4 confirmed: WALR absorbs effect — bank relay is the channel')
        else:
            print('⚠  Direct effect persists — partial exclusion or model misspecification')

    return res

# ══════════════════════════════════════════════════════════════════════════════
# §8  FIGURES
# ══════════════════════════════════════════════════════════════════════════════

def generate_figures(panel, V, vocab, best_k, ml_res):
    print('\n  Generating figures...')
    C = {'hike':'#D85A30','cut':'#378ADD','hold':'#888780',
         'line':'#1D9E75','light':'#EEF4FB'}

    # Fig 1 — Hawkishness time series
    fig, ax = plt.subplots(figsize=(13, 4))
    ax.plot(panel['date'], panel['hawk_score'],
            color=C['line'], lw=2, label='Hawkishness score (LSA → QR → Power Iter)')
    ax.axhline(0, ls='--', color='gray', lw=0.8, alpha=0.5)
    for dec, c in [('hike',C['hike']),('hold',C['hold']),('cut',C['cut'])]:
        sub = panel[panel['decision']==dec]
        ax.scatter(sub['date'], sub['hawk_score'], color=c, s=60,
                   zorder=4, label=dec.capitalize())
    ax.set_title('RBI Hawkishness Score — LSA via QR Decomposition + Power Iteration',
                 fontsize=11)
    ax.set_ylabel('Standardised Score hₜ')
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%b'%y"))
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
    ax.legend(frameon=False, fontsize=9)
    plt.xticks(rotation=45); plt.tight_layout()
    plt.savefig(os.path.join(OUTFIG, 'fig1_hawkishness_timeseries.png'))
    plt.close(); print('    ✅ Fig 1')

    # Fig 2 — Score by decision type
    fig, ax = plt.subplots(figsize=(7, 4))
    for dec, c in [('hike',C['hike']),('hold',C['hold']),('cut',C['cut'])]:
        d = panel[panel['decision']==dec]['hawk_score'].dropna()
        if len(d) > 1:
            ax.hist(d, bins=8, alpha=0.5, color=c,
                    label=f'{dec.capitalize()} (n={len(d)})', density=True)
    ax.set_xlabel('Hawkishness Score hₜ'); ax.set_ylabel('Density')
    ax.set_title('Distribution of Hawkishness Score by Decision Type')
    ax.legend(frameon=False); plt.tight_layout()
    plt.savefig(os.path.join(OUTFIG, 'fig2_score_distributions.png'))
    plt.close(); print('    ✅ Fig 2')

    # Fig 3 — Word loadings (eigenvector interpretation)
    n   = 12
    mean_hike_sc = panel[panel['hike']==1]['hawk_score'].mean() if panel['hike'].sum()>0 else 1
    lod = V[:, best_k] if mean_hike_sc >= 0 else -V[:, best_k]
    top_pos_i = np.argsort(lod)[-n:][::-1]
    top_neg_i = np.argsort(lod)[:n]
    words = list(vocab[top_pos_i]) + list(vocab[top_neg_i])
    vals  = list(lod[top_pos_i]) + list(lod[top_neg_i])
    cols  = [C['hike']]*n + [C['cut']]*n

    fig, ax = plt.subplots(figsize=(8, 7))
    ax.barh(range(len(words)), vals, color=cols, height=0.7)
    ax.set_yticks(range(len(words))); ax.set_yticklabels(words, fontsize=9)
    ax.axvline(0, color='black', lw=0.8)
    ax.set_xlabel('Eigenvector Loading (from Power Iteration on AᵀA)')
    ax.set_title(f'Word Loadings on Hawkishness Axis — Component {best_k+1}\n'
                 f'(Discovered by Power Iteration — not pre-specified)')
    plt.tight_layout()
    plt.savefig(os.path.join(OUTFIG, 'fig3_word_loadings.png'))
    plt.close(); print('    ✅ Fig 3')

    # Fig 4 — Predicted P(hike)
    if ml_res and 'sub' in ml_res:
        sub = ml_res['sub']
        fig, ax = plt.subplots(figsize=(13, 4))
        ax.plot(sub['date'], sub['p_hike'], color=C['hike'], lw=2, label='P(hike)')
        ax.axhline(0.5, ls='--', color='gray', lw=0.8)
        for dec, c, s in [('hike',C['hike'],80),('hold',C['hold'],35),('cut',C['cut'],35)]:
            d = sub[sub['decision']==dec]
            ax.scatter(d['date'], d['p_hike'], color=c, s=s, zorder=4,
                       label=dec.capitalize())
        ax.set_ylabel('P(hike)'); ax.set_ylim(-0.05, 1.05)
        ax.set_title('Predicted Probability of Rate Hike from Hawkishness Score (ML §6)')
        ax.legend(frameon=False, fontsize=9)
        ax.xaxis.set_major_formatter(mdates.DateFormatter("%b'%y"))
        ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
        plt.xticks(rotation=45); plt.tight_layout()
        plt.savefig(os.path.join(OUTFIG, 'fig4_predicted_probability.png'))
        plt.close(); print('    ✅ Fig 4')

    # Fig 5 — Single-peaked preferences illustration
    fig, ax = plt.subplots(figsize=(9, 4))
    agents = [('Retail\nborrowers', 4.0, C['cut']),
              ('SMEs',              4.8, '#6FB0E8'),
              ('Commercial\nbanks', 5.8, C['hold']),
              ('Institutional\ninvestors', 6.5, '#BA7517'),
              ('Foreign\nportfolio', 7.2, C['hike'])]
    xs = np.linspace(2, 10, 400)
    for name, theta, c in agents:
        u  = -((xs - theta)**2)
        u  = (u - u.min()) / (u.max() - u.min())
        ax.plot(xs, u, color=c, lw=2.5, label=f'{name} (θ={theta}%)')
        ax.axvline(theta, color=c, lw=0.7, ls=':')
    r_curr = panel['repo_rate'].iloc[-1] if 'repo_rate' in panel.columns else 6.5
    ax.axvline(r_curr, color='black', lw=1.5, ls='--',
               label=f'Current rate ({r_curr}%)')
    ax.axvline(5.8, color='black', lw=2, alpha=0.2, label='Median θ (5.8%)')
    ax.set_xlabel('Policy Rate r (%)'); ax.set_ylabel('Utility uᵢ(r)')
    ax.set_title('Single-Peaked Preferences (Theory §3.2) — Heterogeneous Agent Types')
    ax.legend(fontsize=8, frameon=False); ax.set_xlim(2, 10)
    plt.tight_layout()
    plt.savefig(os.path.join(OUTFIG, 'fig5_single_peaked_prefs.png'))
    plt.close(); print('    ✅ Fig 5')

# ══════════════════════════════════════════════════════════════════════════════
# §9  EXPORT RESULTS
# ══════════════════════════════════════════════════════════════════════════════

def save_all(panel, h1, h2, h3, h4, ml_res):
    panel.to_csv(os.path.join(OUTTAB, 'master_panel.csv'), index=False)
    summary = {
        'H1_t_stat':      h1.get('t', np.nan),
        'H1_p_value':     h1.get('p', np.nan),
        'H1_mean_hike':   h1.get('mean_hike', np.nan),
        'H1_mean_nonhike':h1.get('mean_nonhike', np.nan),
        'H1_LOO_accuracy':ml_res.get('acc', np.nan),
        'H1_majority_baseline': ml_res.get('maj_base', np.nan),
        'H2_gsec_p':      h2.get('gsec_yield_change',{}).get('p', np.nan),
        'H2_inrusd_p':    h2.get('inrusd_change',{}).get('p', np.nan),
        'H3_best_p':      min([v['p'] for v in h3.values()]) if h3 else np.nan,
        'H4_direct_p':    h4.get('direct_p', np.nan),
        'H4_mediated_p':  h4.get('mediated_p', np.nan),
    }
    pd.DataFrame([summary]).T.rename(columns={0:'value'})\
      .to_csv(os.path.join(OUTRES, 'hypothesis_results.csv'))

    with open(os.path.join(OUTRES, 'results_summary.txt'), 'w') as f:
        f.write('RBI SIGNALING PROJECT — HYPOTHESIS TEST RESULTS\n')
        f.write('='*50 + '\n\n')
        for k, v in summary.items():
            sig = ''
            if 'p_value' in k or '_p' in k:
                if not np.isnan(v):
                    sig = ' ***' if v<0.01 else ' **' if v<0.05 \
                          else ' *' if v<0.10 else ' (ns)'
            f.write(f'{k:<30} {v:.4f}{sig}\n')
        f.write('\n*** p<0.01  ** p<0.05  * p<0.10  (ns) not significant\n')

    print(f'  ✅ master_panel.csv  ({len(panel)} rows × {len(panel.columns)} cols)')
    print(f'  ✅ hypothesis_results.csv')
    print(f'  ✅ results_summary.txt')

# ══════════════════════════════════════════════════════════════════════════════
# MAIN
# ══════════════════════════════════════════════════════════════════════════════

def main():
    print('='*65)
    print('RBI MONETARY POLICY SIGNALING — FULL PIPELINE')
    print('='*65)

    print('\n[§1] Loading rate decision database...')
    decisions = load_rate_decisions()
    print(f'  ✅ {len(decisions)} decisions  '
          f'({decisions["decision"].value_counts().to_dict()})')

    print('\n[§2A] Loading MPC minutes corpus...')
    corpus = load_corpus(decisions)
    print(f'  ✅ {len(corpus)} documents')

    print('\n[§2B] Loading market data...')
    market = load_market_data(decisions)

    print('\n[§3+§4] Running NLP + QR + Power Iteration pipeline...')
    nlp_df, V, vocab, best_k = build_hawkishness_scores(corpus)

    print('\n[§5] Building master panel...')
    panel = build_master_panel(nlp_df, market)

    print('\n[§6] Running ML section...')
    ml_res = run_ml_section(panel)

    print('\n[§7] Running hypothesis tests...')
    h1 = h1_credibility(panel)
    h2 = h2_market_response(panel)
    h3 = h3_bank_relay(panel)
    h4 = h4_retail_exclusion(panel)

    print('\n[§8] Generating figures...')
    generate_figures(panel, V, vocab, best_k, ml_res)

    print('\n[§9] Saving results...')
    save_all(panel, h1, h2, h3, h4, ml_res)

    print('\n' + '='*65)
    print('DONE')
    print('='*65)
    print(f'\n  outputs/figures/  — {len(os.listdir(OUTFIG))} figures')
    print(f'  outputs/tables/   — {len(os.listdir(OUTTAB))} tables')
    print(f'  outputs/results/  — {len(os.listdir(OUTRES))} result files')

if __name__ == '__main__':
    main()
```

```python
# ── PATCH CELL v2 — safe version, no recursion risk ──────────────────────────
# Run this ONCE after the main pipeline cell, before main()

import numpy as np
import pandas as pd
from scipy import stats
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.metrics import accuracy_score, classification_report

def build_master_panel(nlp_df, market_df):
    panel    = market_df.copy()
    panel['date'] = pd.to_datetime(panel['date'])
    nlp_df   = nlp_df.sort_values('date')

    nlp_cols = ['hawk_score','hawk_lag1','hawk_lag2','hawk_lag3',
                'hawk_change','hawk_ma3','dict_score',
                'sentiment','hawk_freq','dove_freq',
                'inflation_freq','growth_freq']

    for col in nlp_cols:
        if col not in nlp_df.columns:
            continue
        vals = []
        for d in panel['date']:
            pre = nlp_df[nlp_df['date'] <= d]
            vals.append(pre[col].iloc[-1] if not pre.empty else np.nan)
        panel[col] = vals

    panel = panel.dropna(subset=['hawk_score']).reset_index(drop=True)

    if panel['hike'].sum() > 0:
        mean_hike    = panel[panel['hike']==1]['hawk_score'].mean()
        mean_nonhike = panel[panel['hike']==0]['hawk_score'].mean()
        if mean_hike < mean_nonhike:
            print('  ↺ Flipping sign: making positive = hawkish')
            for c in ['hawk_score','hawk_lag1','hawk_lag2','hawk_lag3',
                      'hawk_change','hawk_ma3']:
                if c in panel.columns:
                    panel[c] = -panel[c]

    # ── Covid dummy (Fix 2) ───────────────────────────────────────────────────
    panel['covid_dummy'] = (
        (panel['date'] >= '2020-03-01') &
        (panel['date'] <= '2021-12-31')
    ).astype(int)

    print(f'  Master panel: {len(panel)} meetings × {len(panel.columns)} variables')
    print(f'  Covid dummy: {panel["covid_dummy"].sum()} meetings flagged')
    return panel


def run_ml_section(panel):
    print('\n' + '═'*60)
    print('§6 ML SECTION — LOGISTIC REGRESSION + LOO-CV')
    print('═'*60)

    candidate_feats = ['hawk_lag1','hawk_change','sentiment',
                       'inflation_freq','dict_score','cpi_yoy','covid_dummy']
    feats = [f for f in candidate_feats if f in panel.columns]
    sub   = panel[feats + ['hike','date','decision']].dropna()

    if len(sub) < 10:
        print('  ⚠  Insufficient data')
        return {}

    X      = sub[feats].values.astype(float)
    y      = sub['hike'].values
    scaler = StandardScaler()
    X_sc   = scaler.fit_transform(X)

    print(f'\n  Features: {feats}')
    print(f'  Obs: {len(sub)}  |  Hikes: {y.sum()}  |  Non-hikes: {(1-y).sum()}')

    try:
        import statsmodels.api as sm
        X_sm  = sm.add_constant(X_sc)
        logit = sm.Logit(y, X_sm).fit(disp=False)
        print(logit.summary())
        coef_df = pd.DataFrame({
            'feature': ['const'] + feats,
            'coef':    logit.params,
            'p_value': logit.pvalues,
        })
    except Exception:
        X_sm = np.column_stack([np.ones(len(X_sc)), X_sc])
        b, se, t, p = _ols_numpy(X_sm, y.astype(float))
        coef_df = pd.DataFrame({'feature':['const']+feats,'coef':b,'p_value':p})
        print(f'\n  {"Feature":<22} {"Coef":>8} {"p-val":>8}')
        for _, row in coef_df.iterrows():
            sig = '***' if row['p_value']<0.01 else '**' if row['p_value']<0.05 \
                  else '*' if row['p_value']<0.10 else ''
            print(f'  {row["feature"]:<22} {row["coef"]:>+8.3f} {row["p_value"]:>8.4f} {sig}')

    clf   = LogisticRegression(C=0.5, max_iter=500, random_state=42)
    loo   = LeaveOneOut()
    y_hat = cross_val_predict(clf, X_sc, y, cv=loo)
    acc   = accuracy_score(y, y_hat)
    maj   = max(y.mean(), 1 - y.mean())

    print(f'\n  LOO-CV: {acc:.1%}  |  Baseline: {maj:.1%}  |  Lift: {acc-maj:+.1%}')
    print(f'\n{classification_report(y, y_hat, target_names=["Non-hike","Hike"])}')

    clf.fit(X_sc, y)
    probs = clf.predict_proba(X_sc)[:, 1]
    sub   = sub.copy()
    sub['p_hike'] = probs

    return {
        'features': feats, 'X_sc': X_sc, 'y': y,
        'clf': clf, 'scaler': scaler,
        'y_hat': y_hat, 'probs': probs,
        'acc': acc, 'maj_base': maj,
        'coef_df': coef_df, 'sub': sub,
    }


def h1_credibility(panel):
    print('\n' + '═'*60)
    print('H1 — CREDIBILITY (Separating vs Pooling Equilibrium)')
    print('═'*60)

    res = {}
    hike_s    = panel[panel['hike']==1]['hawk_lag1'].dropna()
    nonhike_s = panel[panel['hike']==0]['hawk_lag1'].dropna()

    if len(hike_s) > 1 and len(nonhike_s) > 1:
        t, p = stats.ttest_ind(hike_s, nonhike_s, equal_var=False)
        res  = {'t': t, 'p': p,
                'mean_hike': hike_s.mean(),
                'mean_nonhike': nonhike_s.mean()}
        print(f'\n  [Full sample]')
        print(f'  Mean pre-hike:     {hike_s.mean():+.3f}')
        print(f'  Mean pre-non-hike: {nonhike_s.mean():+.3f}')
        print(f'  t = {t:.3f},  p = {p:.4f}')
        print('  ✅ Credible signal' if p<0.05 else
              '⚠  Marginal' if p<0.10 else '  ❌ Cheap talk')

    # Ex-Covid test
    print(f'\n  [Excluding Covid 2020–2021]')
    panel_ex   = panel[panel['covid_dummy']==0]
    hike_ex    = panel_ex[panel_ex['hike']==1]['hawk_lag1'].dropna()
    nonhike_ex = panel_ex[panel_ex['hike']==0]['hawk_lag1'].dropna()

    if len(hike_ex) > 1 and len(nonhike_ex) > 1:
        t_ex, p_ex = stats.ttest_ind(hike_ex, nonhike_ex, equal_var=False)
        print(f'  Meetings kept: {len(panel_ex)}  |  Hikes: {len(hike_ex)}  |  Non-hikes: {len(nonhike_ex)}')
        print(f'  Mean pre-hike:     {hike_ex.mean():+.3f}')
        print(f'  Mean pre-non-hike: {nonhike_ex.mean():+.3f}')
        print(f'  t = {t_ex:.3f},  p = {p_ex:.4f}')
        if p_ex < 0.05:
            print('  ✅ SIGNIFICANT ex-Covid → Covid was masking the signal')
        elif p_ex < 0.10:
            print('  ⚠  MARGINAL ex-Covid → partial improvement')
        else:
            print('  ❌ Still insignificant → LSA axis needs recalibration')
    return res


def h2_market_response(panel):
    print('\n' + '═'*60)
    print('H2 — DIRECT MARKET RESPONSE (with Covid control)')
    print('═'*60)

    targets = {
        'gsec_yield_change': '10Y G-Sec Yield Change (pp)',
        'inrusd_change':     'INR/USD Change',
    }
    res = {}

    for col, label in targets.items():
        if col not in panel.columns:
            print(f'\n  ⚠  {label}: not available')
            continue

        feats = [f for f in ['hawk_lag1','hawk_change','change_bps','covid_dummy']
                 if f in panel.columns]
        sub   = panel[feats + [col]].dropna()
        if len(sub) < 8:
            continue

        X = np.column_stack([np.ones(len(sub))] +
                            [sub[f].values for f in feats])
        y = sub[col].values.astype(float)
        b, se, t_v, p_v = _ols_numpy(X, y)
        r2 = 1 - np.var(y - X@b) / np.var(y)

        res[col] = {'coef': b[1], 'p': p_v[1], 'r2': r2}
        sig = '***' if p_v[1]<0.01 else '**' if p_v[1]<0.05 \
              else '*' if p_v[1]<0.10 else '(ns)'
        print(f'\n  {label}')
        print(f'    hawk coef = {b[1]:+.4f}  p = {p_v[1]:.4f} {sig}  R² = {r2:.3f}')
        print('    ✅ Credible' if p_v[1]<0.10 else '    ⚠  No significant response')

    return res


def h3_bank_relay(panel):
    print('\n' + '═'*60)
    print('H3 — BANK RELAY (with Covid control)')
    print('═'*60)

    walr_col = next((c for c in panel.columns if c == 'walr_change'), None)
    if walr_col is None:
        print('  ⚠  WALR data not available — H3 skipped')
        return {}

    res = {}
    p   = panel.copy().sort_values('date').reset_index(drop=True)

    for lag in [1, 2, 3]:
        p[f'hawk_l{lag}'] = p['hawk_score'].shift(lag)
        ctrl = [f'hawk_l{lag}','change_bps'] + \
               (['covid_dummy'] if 'covid_dummy' in p.columns else [])
        sub  = p[ctrl + [walr_col]].dropna()
        if len(sub) < 8:
            continue

        X   = np.column_stack([np.ones(len(sub))] +
                               [sub[c].values for c in ctrl])
        y   = sub[walr_col].values.astype(float)
        b, se, t_v, p_v = _ols_numpy(X, y)
        r2  = 1 - np.var(y - X@b) / np.var(y)
        sig = '***' if p_v[1]<0.01 else '**' if p_v[1]<0.05 \
              else '*' if p_v[1]<0.10 else '(ns)'

        res[f'lag{lag}'] = {'coef': b[1], 'p': p_v[1], 'r2': r2}
        print(f'  Lag {lag}: β(hawk)={b[1]:+.4f} p={p_v[1]:.4f} {sig}  '
              f'β(rate)={b[2]:+.4f}  R²={r2:.3f}')

    if res:
        best = min(res, key=lambda k: res[k]['p'])
        print(f'\n  Best lag: {best}  (p={res[best]["p"]:.4f})')
        print('  ✅ Bank relay confirmed' if res[best]['p']<0.10
              else '  ⚠  Weak/no relay evidence')
    return res


def h4_retail_exclusion(panel):
    print('\n' + '═'*60)
    print('H4 — RETAIL EXCLUSION (with Covid control)')
    print('═'*60)

    credit_col = next((c for c in panel.columns if c == 'credit_growth_change'), None)
    walr_col   = next((c for c in panel.columns if c == 'walr_change'), None)

    if credit_col is None:
        print('  ⚠  Retail credit data not available')
        return {}

    base_feats = [f for f in ['hawk_lag1','change_bps','cpi_yoy','covid_dummy']
                  if f in panel.columns]
    sub = panel[base_feats + [credit_col] +
                ([walr_col] if walr_col else [])].dropna()

    if len(sub) < 8:
        print(f'  ⚠  Only {len(sub)} obs — skipping')
        return {}

    y  = sub[credit_col].values.astype(float)
    X1 = np.column_stack([np.ones(len(sub))] +
                          [sub[f].values for f in base_feats])
    b1, _, _, p1 = _ols_numpy(X1, y)
    r2_m1 = 1 - np.var(y - X1@b1) / np.var(y)
    print(f'\n  Model 1 (direct + Covid): hawk_lag1 p={p1[1]:.4f}  R²={r2_m1:.3f}')
    res = {'direct_p': p1[1]}

    if walr_col:
        X2 = np.column_stack([np.ones(len(sub))] +
                              [sub[f].values for f in base_feats + [walr_col]])
        b2, _, _, p2 = _ols_numpy(X2, y)
        r2_m2 = 1 - np.var(y - X2@b2) / np.var(y)
        print(f'  Model 2 (+ WALR + Covid): hawk_lag1 p={p2[1]:.4f}  R²={r2_m2:.3f}')
        res['mediated_p'] = p2[1]
        if p1[1] >= 0.10:
            print('  ✅ H4 confirmed: retail excluded from signal')
        elif p1[1] < 0.10 and p2[1] >= 0.10:
            print('  ✅ H4 confirmed: WALR absorbs effect')
        else:
            print('  ⚠  Direct effect persists')

    return res


print('✅ All patches applied (v2 — no recursion risk)')
print('   Run main() in the next cell')

```
